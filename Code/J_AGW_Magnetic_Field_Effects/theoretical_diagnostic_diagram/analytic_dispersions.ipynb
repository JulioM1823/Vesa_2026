# Solving for the Boundaries of Mihalas & Toomre 1981

```python
# Define the coefficients
cs = sp.Symbol('cs')
kh = sp.Symbol('kh[i]')
N = sp.Symbol('N')
H = sp.Symbol('H')
omega = sp.Symbol('omega[j]')
wac = sp.Symbol('wac')
tau = sp.Symbol('tau')
gamma = sp.Symbol('gamma')
g = sp.Symbol('g')
kz = sp.Symbol('kz')
a = sp.Symbol('a')
ay = sp.Symbol('ay')
az = sp.Symbol('az')
theta = sp.Symbol('theta*np.pi/180')

# Define the equation for Mihalas and Toomre 1981 & Souffrin 1966
eq_mt81 = pow(N, 2)*pow(kh, 2)/pow(omega, 2) + pow(omega, 2)/pow(cs, 2) -1/(4*pow(H, 2)) - pow(kh, 2)
eq_sf66 = (pow(omega, 2) - pow(wac, 2))/pow(cs, 2) - (pow(omega, 2) - pow(N, 2))*pow(kh, 2)/pow(omega, 2)

# Solve the dispersion relation for omega
roots_mt81 = sp.solve(eq_mt81, omega)
roots_sf66 = sp.solve(eq_sf66, omega)

# Print the results in python code
print('Solutions for the Mihalas and Toomre 1981 dispersion relation')
print('-------------------------------------------------------------')
print(sp.python(roots_mt81[0]))
print(sp.python(roots_mt81[1]))
print(sp.python(roots_mt81[2]))
print(sp.python(roots_mt81[3]))
print('\nSolutions for the Souffrin 1966 dispersion relation')
print('---------------------------------------------------')
print(sp.python(roots_sf66[0]))
print(sp.python(roots_sf66[1]))
print(sp.python(roots_sf66[2]))
print(sp.python(roots_sf66[3]))
```

# Solving for the Boundaries of Bunte 1993.

## Non-Magnetic Case

```python
# Define h parameters
hr = (pow(omega, 2) - pow(wac, 2)) / pow(cs, 2) + pow(kh, 2)*(pow(N, 2)/pow(omega, 2) - 1) -(1)/(1 + pow(omega*tau*gamma, 2))*(pow(kh, 2) - pow(omega, 4)/pow(g, 2))*pow(N, 2)/pow(omega, 2)
hi = -omega*tau*gamma/(1 + pow(omega*tau*gamma, 2))*(pow(kh, 2) - pow(omega, 4)/pow(g, 2))*pow(N, 2)/pow(omega, 2)

# Expand hr
hi_exp = sp.expand(hr)

# Find common denominator and simplify
hr_den = sp.simplify(sp.ratsimp(hi_exp))

# Take the numerator
hr_num = sp.numer(hr_den)

# Group by factors of omega
hr_group = sp.collect(hr_num, omega)

# Solve for omega
roots_hr = sp.solve(hr_group, omega)

# Print the results in python code
print('\nSolutions for the general dispersion relation')
print('--------------------------------------------')
print(sp.python(roots_hr[0]))
print(sp.python(roots_hr[1]))
print(sp.python(roots_hr[2]))
print(sp.python(roots_hr[3]))
```

## Magnetic Case

```python
# Define symbol for the alfven speed
c_alf = sp.Symbol('c_alf')

# Define parameterizations
p1 = (cs**2 + c_alf**2)*omega**2 - cs**2*c_alf**2*kh**2
p_gamma = (cs**2 + gamma*c_alf**2)*omega**2 - cs**2*c_alf**2*kh**2
g1 = omega**4 + g**2*(cs**2/(g*H) - 1)*kh**2
g_gamma = gamma*omega**4 + gamma*g**2*(cs**2/(gamma*g*H) - 1)*kh**2
hr = -(1/(4*H**2)) - kh**2 + (p_gamma*g_gamma + (omega*tau*gamma)**2*p1*g1) / (p_gamma**2 + (omega*tau*gamma)**2*p1**2)

# Simplify the expression
hr_simp = sp.simplify(hr)

# Take the numerator
hr_num = sp.numer(hr_simp)

# Group by factors of omega
hr_group = sp.collect(sp.expand(hr_num), omega)

# Access the coefficients of hr_group
coeff0 = hr_group.coeff(omega, 0)
coeff1 = hr_group.coeff(omega, 1)
coeff2 = hr_group.coeff(omega, 2)
coeff3 = hr_group.coeff(omega, 3)
coeff4 = hr_group.coeff(omega, 4)
coeff5 = hr_group.coeff(omega, 5)
coeff6 = hr_group.coeff(omega, 6)
coeff7 = hr_group.coeff(omega, 7)
coeff8 = hr_group.coeff(omega, 8)

# Print the results in python code
print(sp.python(coeff0))
print(sp.python(coeff1))
print(sp.python(coeff2))
print(sp.python(coeff3))
print(sp.python(coeff4))
print(sp.python(coeff5))
print(sp.python(coeff6))
print(sp.python(coeff7))
print(sp.python(coeff8))
```
